Passengerid: Identificador único numérico asignado a cada pasajero.

Age: Edad del pasajero en años.

Fare: Tarifa pagada por el billete/pasaje.

Sex: Género del pasajero, codificado de forma numérica.

sibsp: Número de hermanos o cónyuges a bordo.

Parch: Número de padres o hijos a bordo.

Pclass: Clase del billete o clase social del pasajero.

Embarked: Puerto de embarque codificado de forma numérica.

2urvived: Variable objetivo que indica si el pasajero sobrevivió o no, codificada de forma numérica.
